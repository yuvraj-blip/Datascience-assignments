```python
#otulier detection using IQR method
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1
outlier_condition = (data < (Q1 - 1.5 * IQR)) | (data > (Q3 + 1.5 * IQR))
outliers = data[outlier_condition]
print("Outliers detected using IQR method:")
print(outliers.count())
```

```python
non_outlier_condition = ~((data < (Q1 - 1.5 * IQR)) | (data > (Q3 + 1.5 * IQR))).any(axis=1)

# Trim outlier rows
trimmed_data = data[non_outlier_condition].copy()
```

```python
trimmed_data.info()
```

```python
#correlation matrix
correlation_matrix = trimmed_data.corr()
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot = True, cmap = 'coolwarm', fmt = '.2f')
plt.title('Correlation Matrix')
plt.show()
```

```python
#Standardizing the variable to have a mean = 0 and standard deviation = 1
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data_scaled = scaler.fit_transform(trimmed_data)
data_scaled = pd.DataFrame(data_scaled, columns=trimmed_data.columns)
data_scaled.head()
```

```python
data_scaled.describe()
```

```python
#implementing PCA
from sklearn.decomposition import PCA
pca = PCA()
pca.fit(data_scaled)
```

```python
explained_variance_ratio = pca.explained_variance_ratio_
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(explained_variance_ratio) + 1), explained_variance_ratio, marker='o', linestyle='--', color='blue')
#plt.bar(range(1, len(explained_variance_ratio) + 1), explained_variance_ratio, alpha=0.7, color='blue')
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance Ratio')
plt.title('Scree Plot')
plt.show()

# 4. Cumulative Explained Variance plot
cumulative_explained_variance = explained_variance_ratio.cumsum()
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(cumulative_explained_variance) + 1), cumulative_explained_variance, marker='o', linestyle='--', color='red')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Explained Variance Plot')
plt.grid(True)
plt.show()
```

__According to the scree plot the number of components that we should use 2, since the formation of elbow is after that.__

```python
#pca
pca = PCA(n_components=2)  
pca_result = pca.fit_transform(data_scaled)
pca_result = pd.DataFrame(pca_result, columns=['PC1', 'PC2'])
pca_result.head()
```

```python
#clustering with original data [k means]
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
inertias = []
K = range(1, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(trimmed_data)
    inertias.append(kmeans.inertia_)

plt.plot(K, inertias, marker='o')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal k')
plt.show()
```

```python
#k means with 3 cluters for data_scaled 
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(trimmed_data)

#visualization of clusters
plt.figure(figsize=(10,6))
plt.scatter(trimmed_data.iloc[:, 0], trimmed_data.iloc[:, 1], c=kmeans.labels_, cmap='viridis', marker='o', edgecolor='k', s=50)
plt.title('K-Means Clustering with 3 Clusters')
plt.show()
```

```python
score = silhouette_score(trimmed_data,kmeans.labels_)
print("Silhouette Score: ", score)
```

```python
inertias = []
K = range(1, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(pca_result)
    inertias.append(kmeans.inertia_)

plt.plot(K, inertias, marker='p')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal k')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(pca_result)

#visualization of clusters
plt.figure(figsize=(10,6))
plt.scatter(pca_result.iloc[:, 0], pca_result.iloc[:, 1], c=kmeans.labels_, cmap='viridis', marker='o', edgecolor='k', s=50)
plt.title('K-Means Clustering with 3 Clusters for pca')
plt.show()
```

```python
score = silhouette_score(pca_result,kmeans.labels_)
print("Silhouette Score: ", score)
```

## Task 5: Comparison and Analysis

### 1. Comparison of Clustering Results (Original vs. PCA Data)

- On the original (trimmed and scaled) dataset, K-means clustering with 3 clusters gave a silhouette score of **0.23**.
- On the PCA-transformed data (using the first 2 principal components), K-means clustering with 3 clusters gave a much higher silhouette score of **0.62**.
- Visualizations showed that clusters in the PCA space were more clearly separated compared to the original feature space.

### 2. Similarities and Differences Observed

- Both methods used K-means and the same number of clusters, but the quality of clustering was much better after PCA.
- The original dataset's clusters overlapped more, making it harder to distinguish between them, while PCA-transformed clusters were more compact and well-separated.

### 3. Impact of Dimensionality Reduction on Clustering

- Applying PCA significantly improved clustering performance, as seen from the higher silhouette score and clearer cluster separation in plots.
- PCA reduced noise and redundancy by focusing on the most important features, making the clustering algorithm more effective.

### 4. Trade-offs: PCA vs. Clustering on Original Data

| Aspect                   | Clustering on Original Data | Clustering after PCA        |
|--------------------------|----------------------------|----------------------------|
| Performance (Silhouette) | Lower (0.23)               | Higher (0.62)              |
| Interpretability         | Easier (original features) | Harder (principal components) |
| Computation              | Slower (more features)     | Faster (fewer features)    |
| Cluster Separation       | Less clear                 | More distinct              |

- **Trade-off:** PCA improves clustering quality and efficiency but makes it harder to interpret the clusters in terms of original features.

## Task 6: Conclusion and Insights

### 1. Key Findings and Insights

- Clustering directly on the original dataset resulted in overlapping clusters and a low silhouette score.
- Using PCA for dimensionality reduction before clustering led to much better separation of clusters and a significantly higher silhouette score.
- PCA helped to remove noise and highlight the most important patterns in the data.

### 2. Practical Implications

- Using PCA before clustering is highly beneficial when dealing with high-dimensional data or when features are correlated.
- PCA can simplify the clustering process and improve results, but interpreting clusters becomes less straightforward.

### 3. Recommendations

- **Use PCA before clustering** when your dataset has many features, high correlation, or when initial clustering results are poor.
- **Cluster on original data** if you need to interpret clusters using the original features or if your data is already low-dimensional and well-separated.
- Always check clustering quality (e.g., silhouette score) and visualize results to guide your approach.

---

**In summary:**  
PCA can greatly enhance clustering performance by reducing dimensionality and noise, but it comes at the cost of interpretability. Use PCA as a preprocessing step when clustering quality is more important than feature-level interpretation.
