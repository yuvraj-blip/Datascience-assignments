```python
import numpy as np
import scipy.stats as stats
import math


data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48,
        1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
sample_mean = np.mean(data)
# ddof=1 is important to get Sample Std Dev (dividing by n-1, not n)
sample_std = np.std(data, ddof=1)

print(f"Sample Size (n): {n}")
print(f"Sample Mean: {sample_mean:.4f}")
print(f"Sample Std Dev: {sample_std:.4f}")
print("-" * 30)

# --- TASK A: 99% CI using T-Distribution (Unknown Sigma) ---
# We use T-Distribution because sample size is small (n < 30)
confidence_level = 0.99
df = n - 1 # Degrees of Freedom

# Calculate Interval
# loc = Mean, scale = Standard Error (std / sqrt(n))
ci_t = stats.t.interval(confidence_level, df, loc=sample_mean, scale=sample_std/math.sqrt(n))

print(f"\n Task A (T-Distribution)\n ")
print(f"Rationale: n < 30.")
print(f"99% Confidence Interval: {ci_t}")
print(f"Formatted: ({ci_t[0]:.4f}, {ci_t[1]:.4f})")

# --- TASK B: 99% CI using Z-Distribution
# We use Z-Distribution because Population Std Dev is KNOWN.
sigma = 0.2

# Calculate Interval
# scale = Standard Error (sigma / sqrt(n))
ci_z = stats.norm.interval(confidence_level, loc=sample_mean, scale=sigma/math.sqrt(n))

print(f"\n--- Task B (Z-Distribution)")
print(f"Rationale: Population Sigma is known (0.2).")
print(f"99% Confidence Interval: {ci_z}")
print(f"Formatted: ({ci_z[0]:.4f}, {ci_z[1]:.4f})")
```
