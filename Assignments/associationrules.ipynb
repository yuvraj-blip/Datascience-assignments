```python
transaction = df[0].dropna().apply(lambda x: x.split(','))

transaction = transaction.tolist()

transaction[:10]
```

```python
te = TransactionEncoder()
te_ary = te.fit(transaction).transform(transaction)
basket_df = pd.DataFrame(te_ary, columns=te.columns_)
basket_df
```

```python
import warnings
warnings.filterwarnings('ignore', category=DeprecationWarning, module='jupyter_client')
```

```python
from mlxtend.frequent_patterns import apriori

frequent_itemsets = apriori(basket_df, min_support=0.02, use_colnames=True)
print(frequent_itemsets)
```

```python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
display(rules)
```

Based on the association rules provided, here is an analysis of the purchasing patterns and customer behavior insights.

### 1. High-Level Summary

The dataset reveals two distinct types of purchasing behaviors: **Meal Preparation** (ingredients bought together to make a specific dish) and **Staple Accumulation** (essential items added to the basket regardless of the main purchase).

* **Most Influential Item:** `Spaghetti` (The anchor for meal-specific rules).
* **Most Common Accompaniment:** `Mineral Water` (The universal add-on).
* **Strongest Association:** `Ground Beef`  `Spaghetti` (Lift: 2.29).

---

### 2. Key Pattern Clusters

#### A. The "Pasta Night" Phenomenon (Strongest Correlation)

The strongest rules in your entire dataset revolve around cooking pasta. The data suggests that when customers buy specific proteins or sauces, they are almost certainly making a pasta dish.

* **The Rules:**
* **Rule 64:** `Ground Beef`  `Spaghetti` (**Lift: 2.29**, Confidence: ~40%)
* **Rule 86:** `Olive Oil`  `Spaghetti` (**Lift: 2.00**, Confidence: ~35%)
* **Rule 92:** `Tomatoes`  `Spaghetti` (**Lift: 1.76**, Confidence: ~31%)
* **Rule 90:** `Shrimp`  `Spaghetti` (**Lift: 1.70**, Confidence: ~30%)


* **Insight:** "Spaghetti" is not just a product; it is a destination category. Customers buying ground beef, olive oil, or tomatoes are statistically very likely to be planning a specific meal (Bolognese or Seafood Pasta).
* **Actionable Advice:** Place parmesan cheese, garlic bread, or basil stations directly next to the Ground Beef and Fresh Tomato sections, not just in the pasta aisle.

#### B. The "Mineral Water" Default

`Mineral Water` appears as a consequent in a massive number of rules (Burgers, Cake, Chicken, Chocolate, Pancakes, etc.).

* **The Rules:**
* **Rule 78:** `Soup`  `Mineral Water` (**Lift: 1.91**)
* **Rule 54:** `Frozen Vegetables`  `Mineral Water` (**Lift: 1.57**)
* **Rule 11:** `Chicken`  `Mineral Water` (**Lift: 1.59**)


* **Insight:** Mineral water behaves as a "basket filler." It has high support but is often bought alongside unrelated items. The high association with **Soup** (Lift 1.91) is particularly interesting—it suggests a specific "Health/Light Meal" shopper profile who values hydration and light food.

#### C. The "Heavy Breakfast" or "Keto" Shopper

There is a surprisingly strong correlation between Burgers and Eggs.

* **The Rule:**
* **Rule 0:** `Burgers`  `Eggs` (**Lift: 1.84**)


* **Insight:** While eggs are typically breakfast and burgers are dinner, their strong association suggests a specific dietary pattern (potentially high-protein/low-carb diets) or a household purchasing distinct protein staples simultaneously.

#### D. The "Creamy Casserole" Pattern

* **The Rule:**
* **Rule 52:** `Frozen Vegetables`  `Milk` (**Lift: 1.91**)


* **Insight:** This is a very specific cooking behavior. The high lift indicates that when people buy frozen veggies, they are frequently buying milk, likely to create sauces, soups, or casseroles.

---

### 3. Metric-Based Interpretation

To understand *why* these rules matter, we look at the three key metrics:

| Rule Pair | Metric Focus | Interpretation |
| --- | --- | --- |
| **Ground Beef  Spaghetti** | **High Lift (2.29)** | **This is a dominant strategy.** The purchase of beef *drastically* increases the probability of buying spaghetti. These items are interdependent. |
| **Mineral Water  Whole Wheat Rice** | **Low Confidence (8%)** | **Weak prediction.** Even though the Lift is positive (1.44), knowing someone bought water only gives you an 8% certainty they will buy rice. This is not a strong rule for promotions. |
| **Burgers  Eggs** | **High Confidence (33%)** | **Reliable habit.** 1 in 3 people who buy burgers also buy eggs. This is a safe bet for bundle pricing. |

### 4. Business Recommendations

1. **Dynamic Bundling:** Create a **"Sunday Dinner Bundle"** offering a discount if Ground Beef, Spaghetti, and Tomato Sauce are bought together. The data proves these naturally go together.
2. **Store Layout:**
* **Cross-Merchandising:** Move a display of `Spaghetti` to the `Meat` department (specifically near Ground Beef).
* **Impulse Buys:** Since `Mineral Water` is bought with *everything* (Cake, Chicken, Pancakes), place single-serve water bottles at the checkout of every aisle, not just the beverage aisle.


3. **Inventory Planning:** If you run a promotion on `Olive Oil`, expect a surge in `Spaghetti` sales. Ensure stock levels for pasta are adjusted up when oil or beef is discounted.
